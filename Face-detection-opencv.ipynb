```python
from matplotlib import pyplot as plt
import cv2
from google.colab import files
import urllib.request
```

```python
uploaded_image=files.upload()
```

```python
# Download a pre-trained Haar Cascade XML for face detection
# This is a common front face detector provided by OpenCV
haar_cascade_url = 'https://raw.githubusercontent.com/opencv/opencv/4.x/data/haarcascades/haarcascade_frontalface_default.xml'
haar_cascade_filename = 'haarcascade_frontalface_default.xml'
urllib.request.urlretrieve(haar_cascade_url, haar_cascade_filename)

print(f"Downloaded {haar_cascade_filename}")
```

```python
image=cv2.imread(list(uploaded_image.keys())[0])
image=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
```

```python
plt.imshow(image)
```

```python
model = cv2.CascadeClassifier("haarcascade_frontalface_default.xml")
```

```python
face=model.detectMultiScale(image)
```

```python
print(face)
```

```python
image.shape
```

```python
x=face[0][0]
y=face[0][1]
a=face[0][2]
b=face[0][3]

image=cv2.rectangle(image,(x,y),(x+a,y+b),(0,255,0),4)
image=cv2.cvtColor(image,cv2.COLOR_BGR2RGB)
plt.imshow(image)
```
